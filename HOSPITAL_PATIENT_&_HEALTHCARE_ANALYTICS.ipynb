```python
# ============================================================
#  HOSPITAL PATIENT & HEALTHCARE ANALYTICS
# Complete Portfolio Data Analytics Project
# ============================================================

# ============================================================
# PROJECT INFORMATION
# ============================================================

PROJECT_TITLE = """
Hospital Patient & Healthcare Analytics
Exploring Patient Demographics, Admissions, Treatment Patterns,
Healthcare Utilization and Billing
"""

print(PROJECT_TITLE)


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import chi2_contingency
from scipy.stats import ttest_ind

from google.colab import files

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (10, 6)

print("✅ Libraries imported successfully.")


# ============================================================
# 2. UPLOAD DATASET
# ============================================================

print("\n📂 Upload your hospital healthcare CSV file")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("\n✅ Dataset loaded successfully")
print("File:", file_name)


# ============================================================
# 3. EXPECTED COLUMNS
# ============================================================

expected_columns = [
    "Age",
    "Gender",
    "Medical Condition",
    "Date of Admission",
    "Hospital",
    "Insurance Provider",
    "Billing Amount",
    "Admission Type",
    "Discharge Date",
    "Medication",
    "Test Results"
]

print("\n========== EXPECTED COLUMNS ==========")

print(expected_columns)


# ============================================================
# 4. CHECK COLUMNS
# ============================================================

print("\n========== DATASET COLUMNS ==========")

print(df.columns.tolist())

missing_columns = [
    column
    for column in expected_columns
    if column not in df.columns
]

extra_columns = [
    column
    for column in df.columns
    if column not in expected_columns
]

if missing_columns:

    print("\n⚠️ Missing expected columns:")
    print(missing_columns)

else:

    print("\n✅ All expected columns are present.")

if extra_columns:

    print("\nAdditional columns found:")
    print(extra_columns)


# ============================================================
# 5. DATASET SIZE
# ============================================================

print("\n========== DATASET SIZE ==========")

rows, columns = df.shape

print("Rows:", rows)
print("Columns:", columns)


# ============================================================
# 6. FIRST RECORDS
# ============================================================

print("\n========== FIRST 10 RECORDS ==========")

display(df.head(10))


# ============================================================
# 7. RANDOM SAMPLE
# ============================================================

print("\n========== RANDOM SAMPLE ==========")

display(df.sample(
    min(10, len(df)),
    random_state=42
))


# ============================================================
# 8. DATA TYPES
# ============================================================

print("\n========== DATA TYPES ==========")

display(
    df.dtypes.to_frame("Data Type")
)


# ============================================================
# 9. DATASET INFORMATION
# ============================================================

print("\n========== DATASET INFORMATION ==========")

df.info()


# ============================================================
# 10. MISSING VALUES
# ============================================================

print("\n========== MISSING VALUES ==========")

missing = df.isnull().sum()

missing_percentage = (
    df.isnull().mean() * 100
)

missing_table = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": missing.values,
    "Missing Percentage": missing_percentage.values
})

display(
    missing_table
    .sort_values(
        "Missing Values",
        ascending=False
    )
)


# ============================================================
# 11. DUPLICATES
# ============================================================

print("\n========== DUPLICATE RECORDS ==========")

duplicate_count = df.duplicated().sum()

print(
    "Duplicate rows:",
    duplicate_count
)


# ============================================================
# 12. UNIQUE VALUES
# ============================================================

print("\n========== UNIQUE VALUES ==========")

unique_table = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [
        df[column].nunique()
        for column in df.columns
    ]
})

display(unique_table)


# ============================================================
# 13. DATA CLEANING
# ============================================================

print("\n========== DATA CLEANING ==========")

# Remove duplicate rows
before_duplicates = len(df)

df = df.drop_duplicates().copy()

after_duplicates = len(df)

print(
    "Duplicates removed:",
    before_duplicates - after_duplicates
)


# Convert date columns
df["Date of Admission"] = pd.to_datetime(
    df["Date of Admission"],
    errors="coerce"
)

df["Discharge Date"] = pd.to_datetime(
    df["Discharge Date"],
    errors="coerce"
)


# Convert numeric columns
df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

df["Billing Amount"] = pd.to_numeric(
    df["Billing Amount"],
    errors="coerce"
)


# Clean text columns
text_columns = [
    "Gender",
    "Medical Condition",
    "Hospital",
    "Insurance Provider",
    "Admission Type",
    "Medication",
    "Test Results"
]

for column in text_columns:

    df[column] = (
        df[column]
        .astype("string")
        .str.strip()
    )


print("\n✅ Basic cleaning completed.")


# ============================================================
# 14. CREATE LENGTH OF STAY
# ============================================================

df["Length of Stay"] = (
    df["Discharge Date"] -
    df["Date of Admission"]
).dt.days


# ============================================================
# 15. CHECK INVALID LENGTH OF STAY
# ============================================================

invalid_stay = (
    df["Length of Stay"] < 0
).sum()

print(
    "\nInvalid negative length-of-stay records:",
    invalid_stay
)


# Remove impossible negative values
df.loc[
    df["Length of Stay"] < 0,
    "Length of Stay"
] = np.nan


# ============================================================
# 16. FINAL MISSING VALUE CHECK
# ============================================================

print("\n========== FINAL MISSING VALUE CHECK ==========")

final_missing = df.isnull().sum()

display(
    final_missing.to_frame(
        "Missing Values"
    )
)


# ============================================================
# 17. DESCRIPTIVE STATISTICS
# ============================================================

print("\n========== DESCRIPTIVE STATISTICS ==========")

display(
    df[
        [
            "Age",
            "Billing Amount",
            "Length of Stay"
        ]
    ].describe().T
)


# ============================================================
# 18. EXECUTIVE KPI SUMMARY
# ============================================================

total_patients = len(df)

average_age = df["Age"].mean()

median_age = df["Age"].median()

average_billing = df["Billing Amount"].mean()

median_billing = df["Billing Amount"].median()

total_billing = df["Billing Amount"].sum()

average_stay = df["Length of Stay"].mean()

median_stay = df["Length of Stay"].median()

unique_hospitals = df["Hospital"].nunique()

unique_conditions = df["Medical Condition"].nunique()

unique_insurance = df["Insurance Provider"].nunique()

unique_medications = df["Medication"].nunique()


print("\n")
print("=" * 70)
print("🏥 HEALTHCARE ANALYTICS — EXECUTIVE KPI SUMMARY")
print("=" * 70)

print(
    f"\nTotal Patient Records       : {total_patients:,}"
)

print(
    f"Average Patient Age        : {average_age:.2f}"
)

print(
    f"Median Patient Age         : {median_age:.2f}"
)

print(
    f"Average Billing Amount     : {average_billing:,.2f}"
)

print(
    f"Median Billing Amount      : {median_billing:,.2f}"
)

print(
    f"Total Recorded Billing     : {total_billing:,.2f}"
)

print(
    f"Average Length of Stay     : {average_stay:.2f} days"
)

print(
    f"Median Length of Stay      : {median_stay:.2f} days"
)

print(
    f"Number of Hospitals        : {unique_hospitals:,}"
)

print(
    f"Medical Conditions         : {unique_conditions:,}"
)

print(
    f"Insurance Providers        : {unique_insurance:,}"
)

print(
    f"Medications                : {unique_medications:,}"
)

print("=" * 70)


# ============================================================
# 19. PATIENT AGE DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True
)

plt.title(
    "Patient Age Distribution",
    fontsize=15
)

plt.xlabel("Age")

plt.ylabel(
    "Number of Patients"
)

plt.tight_layout()

plt.show()


# ============================================================
# 20. AGE GROUPS
# ============================================================

df["Age Group"] = pd.cut(
    df["Age"],
    bins=[
        0,
        18,
        30,
        45,
        60,
        100
    ],
    labels=[
        "0-18",
        "19-30",
        "31-45",
        "46-60",
        "61+"
    ]
)

age_group_counts = (
    df["Age Group"]
    .value_counts()
    .sort_index()
)

print("\n========== PATIENTS BY AGE GROUP ==========")

display(
    age_group_counts.to_frame(
        "Patients"
    )
)


plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Age Group",
    order=[
        "0-18",
        "19-30",
        "31-45",
        "46-60",
        "61+"
    ]
)

plt.title(
    "Patient Distribution by Age Group"
)

plt.xlabel("Age Group")

plt.ylabel(
    "Number of Patients"
)

plt.tight_layout()

plt.show()


# ============================================================
# 21. GENDER ANALYSIS
# ============================================================

gender_counts = (
    df["Gender"]
    .value_counts()
)

print("\n========== GENDER DISTRIBUTION ==========")

display(
    gender_counts.to_frame(
        "Patients"
    )
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Gender"
)

plt.title(
    "Patient Distribution by Gender"
)

plt.xlabel("Gender")

plt.ylabel(
    "Number of Patients"
)

plt.tight_layout()

plt.show()


# ============================================================
# 22. MEDICAL CONDITIONS
# ============================================================

condition_counts = (
    df["Medical Condition"]
    .value_counts()
)

print("\n========== MEDICAL CONDITIONS ==========")

display(
    condition_counts.to_frame(
        "Patients"
    )
)


plt.figure(figsize=(10, 6))

sns.barplot(
    x=condition_counts.values,
    y=condition_counts.index
)

plt.title(
    "Patient Records by Medical Condition"
)

plt.xlabel(
    "Number of Patients"
)

plt.ylabel(
    "Medical Condition"
)

plt.tight_layout()

plt.show()


# ============================================================
# 23. MEDICAL CONDITION BY GENDER
# ============================================================

condition_gender = pd.crosstab(
    df["Medical Condition"],
    df["Gender"]
)

print(
    "\n========== MEDICAL CONDITION BY GENDER =========="
)

display(condition_gender)


plt.figure(figsize=(11, 6))

condition_gender.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title(
    "Medical Conditions by Gender"
)

plt.xlabel(
    "Medical Condition"
)

plt.ylabel(
    "Number of Patients"
)

plt.xticks(rotation=20)

plt.tight_layout()

plt.show()


# ============================================================
# 24. ADMISSION TYPE
# ============================================================

admission_counts = (
    df["Admission Type"]
    .value_counts()
)

print(
    "\n========== ADMISSION TYPES =========="
)

display(
    admission_counts.to_frame(
        "Patients"
    )
)


plt.figure(figsize=(9, 6))

sns.countplot(
    data=df,
    x="Admission Type"
)

plt.title(
    "Patient Distribution by Admission Type"
)

plt.xlabel(
    "Admission Type"
)

plt.ylabel(
    "Number of Patients"
)

plt.tight_layout()

plt.show()


# ============================================================
# 25. ADMISSION TYPE PERCENTAGE
# ============================================================

admission_percentage = (
    df["Admission Type"]
    .value_counts(
        normalize=True
    ) * 100
)

print(
    "\n========== ADMISSION TYPE PERCENTAGE =========="
)

display(
    admission_percentage
    .round(2)
    .to_frame(
        "Percentage"
    )
)


# ============================================================
# 26. ADMISSION TREND BY MONTH
# ============================================================

df["Admission Month"] = (
    df["Date of Admission"]
    .dt.to_period("M")
)

monthly_admissions = (
    df.groupby(
        "Admission Month"
    )
    .size()
)

monthly_admissions.index = (
    monthly_admissions.index
    .astype(str)
)

print(
    "\n========== MONTHLY ADMISSIONS =========="
)

display(
    monthly_admissions
    .to_frame(
        "Admissions"
    )
)


plt.figure(figsize=(14, 6))

plt.plot(
    monthly_admissions.index,
    monthly_admissions.values,
    marker="o",
    linewidth=2
)

plt.title(
    "Monthly Hospital Admission Trend"
)

plt.xlabel("Month")

plt.ylabel(
    "Number of Admissions"
)

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()


# ============================================================
# 27. LENGTH OF STAY DISTRIBUTION
# ============================================================

print(
    "\n========== LENGTH OF STAY STATISTICS =========="
)

display(
    df["Length of Stay"]
    .describe()
    .to_frame(
        "Length of Stay"
    )
)


plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Length of Stay",
    bins=20,
    kde=True
)

plt.title(
    "Distribution of Patient Length of Stay"
)

plt.xlabel(
    "Length of Stay (Days)"
)

plt.ylabel(
    "Number of Patients"
)

plt.tight_layout()

plt.show()


# ============================================================
# 28. LENGTH OF STAY BY ADMISSION TYPE
# ============================================================

stay_by_admission = (
    df.groupby(
        "Admission Type"
    )["Length of Stay"]
    .agg([
        "count",
        "mean",
        "median",
        "std"
    ])
    .round(2)
)

print(
    "\n========== LENGTH OF STAY BY ADMISSION TYPE =========="
)

display(stay_by_admission)


plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Admission Type",
    y="Length of Stay"
)

plt.title(
    "Length of Stay by Admission Type"
)

plt.xlabel(
    "Admission Type"
)

plt.ylabel(
    "Length of Stay (Days)"
)

plt.tight_layout()

plt.show()


# ============================================================
# 29. BILLING DISTRIBUTION
# ============================================================

print(
    "\n========== BILLING STATISTICS =========="
)

display(
    df["Billing Amount"]
    .describe()
    .to_frame(
        "Billing Amount"
    )
)


plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Billing Amount",
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Hospital Billing Amounts"
)

plt.xlabel(
    "Billing Amount"
)

plt.ylabel(
    "Number of Patients"
)

plt.tight_layout()

plt.show()


# ============================================================
# 30. BILLING BY MEDICAL CONDITION
# ============================================================

billing_condition = (
    df.groupby(
        "Medical Condition"
    )["Billing Amount"]
    .agg([
        "count",
        "mean",
        "median",
        "sum"
    ])
    .sort_values(
        "mean",
        ascending=False
    )
    .round(2)
)

print(
    "\n========== BILLING BY MEDICAL CONDITION =========="
)

display(billing_condition)


plt.figure(figsize=(11, 6))

sns.barplot(
    data=billing_condition.reset_index(),
    x="mean",
    y="Medical Condition"
)

plt.title(
    "Average Billing Amount by Medical Condition"
)

plt.xlabel(
    "Average Billing Amount"
)

plt.ylabel(
    "Medical Condition"
)

plt.tight_layout()

plt.show()


# ============================================================
# 31. BILLING BY ADMISSION TYPE
# ============================================================

billing_admission = (
    df.groupby(
        "Admission Type"
    )["Billing Amount"]
    .agg([
        "count",
        "mean",
        "median",
        "std"
    ])
    .round(2)
)

print(
    "\n========== BILLING BY ADMISSION TYPE =========="
)

display(billing_admission)


plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Admission Type",
    y="Billing Amount"
)

plt.title(
    "Billing Amount by Admission Type"
)

plt.xlabel(
    "Admission Type"
)

plt.ylabel(
    "Billing Amount"
)

plt.tight_layout()

plt.show()


# ============================================================
# 32. INSURANCE PROVIDER ANALYSIS
# ============================================================

insurance_counts = (
    df["Insurance Provider"]
    .value_counts()
)

print(
    "\n========== INSURANCE PROVIDERS =========="
)

display(
    insurance_counts.to_frame(
        "Patients"
    )
)


plt.figure(figsize=(11, 6))

sns.barplot(
    x=insurance_counts.values,
    y=insurance_counts.index
)

plt.title(
    "Patients by Insurance Provider"
)

plt.xlabel(
    "Number of Patients"
)

plt.ylabel(
    "Insurance Provider"
)

plt.tight_layout()

plt.show()


# ============================================================
# 33. BILLING BY INSURANCE PROVIDER
# ============================================================

insurance_billing = (
    df.groupby(
        "Insurance Provider"
    )["Billing Amount"]
    .agg([
        "count",
        "mean",
        "median",
        "sum"
    ])
    .sort_values(
        "mean",
        ascending=False
    )
    .round(2)
)

print(
    "\n========== BILLING BY INSURANCE PROVIDER =========="
)

display(insurance_billing)


plt.figure(figsize=(11, 6))

sns.barplot(
    data=insurance_billing.reset_index(),
    x="mean",
    y="Insurance Provider"
)

plt.title(
    "Average Billing Amount by Insurance Provider"
)

plt.xlabel(
    "Average Billing Amount"
)

plt.ylabel(
    "Insurance Provider"
)

plt.tight_layout()

plt.show()


# ============================================================
# 34. MEDICATION ANALYSIS
# ============================================================

medication_counts = (
    df["Medication"]
    .value_counts()
)

print(
    "\n========== MEDICATION DISTRIBUTION =========="
)

display(
    medication_counts.to_frame(
        "Patients"
    )
)


plt.figure(figsize=(10, 6))

sns.barplot(
    x=medication_counts.values,
    y=medication_counts.index
)

plt.title(
    "Medication Distribution"
)

plt.xlabel(
    "Number of Patients"
)

plt.ylabel(
    "Medication"
)

plt.tight_layout()

plt.show()


# ============================================================
# 35. MEDICATION BY MEDICAL CONDITION
# ============================================================

condition_medication = pd.crosstab(
    df["Medical Condition"],
    df["Medication"]
)

print(
    "\n========== MEDICATION BY MEDICAL CONDITION =========="
)

display(condition_medication)


plt.figure(figsize=(12, 7))

sns.heatmap(
    condition_medication,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Medical Condition vs Medication"
)

plt.xlabel(
    "Medication"
)

plt.ylabel(
    "Medical Condition"
)

plt.tight_layout()

plt.show()


# ============================================================
# 36. TEST RESULTS
# ============================================================

test_results = (
    df["Test Results"]
    .value_counts()
)

print(
    "\n========== TEST RESULTS =========="
)

display(
    test_results.to_frame(
        "Patients"
    )
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Test Results"
)

plt.title(
    "Distribution of Test Results"
)

plt.xlabel(
    "Test Result"
)

plt.ylabel(
    "Number of Patients"
)

plt.tight_layout()

plt.show()


# ============================================================
# 37. MEDICAL CONDITION VS TEST RESULTS
# ============================================================

condition_test = pd.crosstab(
    df["Medical Condition"],
    df["Test Results"]
)

print(
    "\n========== MEDICAL CONDITION VS TEST RESULTS =========="
)

display(condition_test)


plt.figure(figsize=(11, 7))

sns.heatmap(
    condition_test,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Medical Condition vs Test Results"
)

plt.xlabel(
    "Test Result"
)

plt.ylabel(
    "Medical Condition"
)

plt.tight_layout()

plt.show()


# ============================================================
# 38. HOSPITAL ANALYSIS
# ============================================================

hospital_counts = (
    df["Hospital"]
    .value_counts()
)

print(
    "\n========== HOSPITAL ANALYSIS =========="
)

print(
    "Unique hospitals:",
    len(hospital_counts)
)

print(
    "\nTop 10 hospitals by patient records:"
)

display(
    hospital_counts
    .head(10)
    .to_frame(
        "Patients"
    )
)


plt.figure(figsize=(11, 6))

sns.barplot(
    x=hospital_counts.head(10).values,
    y=hospital_counts.head(10).index
)

plt.title(
    "Top 10 Hospitals by Patient Records"
)

plt.xlabel(
    "Number of Patients"
)

plt.ylabel(
    "Hospital"
)

plt.tight_layout()

plt.show()


# ============================================================
# 39. AGE VS BILLING
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Billing Amount",
    alpha=0.5
)

plt.title(
    "Patient Age vs Billing Amount"
)

plt.xlabel("Age")

plt.ylabel(
    "Billing Amount"
)

plt.tight_layout()

plt.show()


age_billing_corr = (
    df[
        [
            "Age",
            "Billing Amount"
        ]
    ]
    .corr()
)

print(
    "\n========== AGE VS BILLING CORRELATION =========="
)

display(
    age_billing_corr.round(3)
)


# ============================================================
# 40. LENGTH OF STAY VS BILLING
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Length of Stay",
    y="Billing Amount",
    alpha=0.5
)

plt.title(
    "Length of Stay vs Billing Amount"
)

plt.xlabel(
    "Length of Stay (Days)"
)

plt.ylabel(
    "Billing Amount"
)

plt.tight_layout()

plt.show()


stay_billing_corr = (
    df[
        [
            "Length of Stay",
            "Billing Amount"
        ]
    ]
    .corr()
)

print(
    "\n========== LENGTH OF STAY VS BILLING CORRELATION =========="
)

display(
    stay_billing_corr.round(3)
)


# ============================================================
# 41. CORRELATION MATRIX
# ============================================================

numeric_columns = [
    "Age",
    "Billing Amount",
    "Length of Stay"
]

correlation_matrix = (
    df[numeric_columns]
    .corr()
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Matrix"
)

plt.tight_layout()

plt.show()


# ============================================================
# 42. CHI-SQUARE TEST FUNCTION
# ============================================================

def chi_square_test(
    data,
    column1,
    column2
):

    table = pd.crosstab(
        data[column1],
        data[column2]
    )

    chi2, p_value, dof, expected = (
        chi2_contingency(table)
    )

    return {
        "Variable 1": column1,
        "Variable 2": column2,
        "Chi-Square": chi2,
        "p-value": p_value,
        "Significant": (
            "Yes"
            if p_value < 0.05
            else "No"
        )
    }


# ============================================================
# 43. CHI-SQUARE TESTS
# ============================================================

chi_results = []

tests = [
    (
        "Gender",
        "Medical Condition"
    ),
    (
        "Gender",
        "Test Results"
    ),
    (
        "Admission Type",
        "Test Results"
    ),
    (
        "Admission Type",
        "Medical Condition"
    ),
    (
        "Insurance Provider",
        "Admission Type"
    ),
    (
        "Medical Condition",
        "Medication"
    )
]

for column1, column2 in tests:

    result = chi_square_test(
        df,
        column1,
        column2
    )

    chi_results.append(result)


chi_results_df = pd.DataFrame(
    chi_results
)

chi_results_df["Chi-Square"] = (
    chi_results_df[
        "Chi-Square"
    ].round(4)
)

chi_results_df["p-value"] = (
    chi_results_df[
        "p-value"
    ].round(6)
)

print(
    "\n========== CHI-SQUARE TEST RESULTS =========="
)

display(
    chi_results_df.sort_values(
        "p-value"
    )
)


# ============================================================
# 44. BILLING GROUP COMPARISON
# ============================================================

print(
    "\n========== BILLING GROUP ANALYSIS =========="
)

admission_groups = (
    df["Admission Type"]
    .dropna()
    .unique()
)

print(
    "Admission groups:",
    admission_groups
)


if len(admission_groups) == 2:

    group1 = df[
        df["Admission Type"]
        == admission_groups[0]
    ]["Billing Amount"]

    group2 = df[
        df["Admission Type"]
        == admission_groups[1]
    ]["Billing Amount"]

    t_stat, p_value = ttest_ind(
        group1,
        group2,
        equal_var=False
    )

    print(
        "\nT-statistic:",
        round(t_stat, 4)
    )

    print(
        "p-value:",
        p_value
    )

    if p_value < 0.05:

        print(
            "\nResult: There is evidence of a "
            "difference in mean billing amounts "
            "between the two admission groups."
        )

    else:

        print(
            "\nResult: There is insufficient evidence "
            "of a difference in mean billing amounts "
            "between the two admission groups."
        )

else:

    print(
        "\nMore than two admission groups exist."
    )

    print(
        "For a full comparison across all groups, "
        "ANOVA would be more appropriate."
    )


# ============================================================
# 45. TOP FINDINGS
# ============================================================

print("\n")
print("=" * 70)
print("🔎 AUTOMATED KEY FINDINGS")
print("=" * 70)


# Most common medical condition
top_condition = (
    df["Medical Condition"]
    .value_counts()
    .idxmax()
)

top_condition_count = (
    df["Medical Condition"]
    .value_counts()
    .max()
)

print(
    f"\n1. Most frequently recorded medical condition:"
)

print(
    f"   {top_condition} "
    f"({top_condition_count:,} records)"
)


# Most common admission
top_admission = (
    df["Admission Type"]
    .value_counts()
    .idxmax()
)

top_admission_count = (
    df["Admission Type"]
    .value_counts()
    .max()
)

print(
    f"\n2. Most common admission type:"
)

print(
    f"   {top_admission} "
    f"({top_admission_count:,} records)"
)


# Most common medication
top_medication = (
    df["Medication"]
    .value_counts()
    .idxmax()
)

top_medication_count = (
    df["Medication"]
    .value_counts()
    .max()
)

print(
    f"\n3. Most frequently recorded medication:"
)

print(
    f"   {top_medication} "
    f"({top_medication_count:,} records)"
)


# Most common test result
top_test_result = (
    df["Test Results"]
    .value_counts()
    .idxmax()
)

top_test_result_count = (
    df["Test Results"]
    .value_counts()
    .max()
)

print(
    f"\n4. Most common test result:"
)

print(
    f"   {top_test_result} "
    f"({top_test_result_count:,} records)"
)


# Highest billing condition
highest_billing_condition = (
    billing_condition["mean"]
    .idxmax()
)

highest_billing_value = (
    billing_condition["mean"]
    .max()
)

print(
    f"\n5. Medical condition with highest "
    f"average billing:"
)

print(
    f"   {highest_billing_condition} "
    f"({highest_billing_value:,.2f})"
)


# Highest billing insurance
highest_insurance = (
    insurance_billing["mean"]
    .idxmax()
)

highest_insurance_value = (
    insurance_billing["mean"]
    .max()
)

print(
    f"\n6. Insurance provider with highest "
    f"average billing:"
)

print(
    f"   {highest_insurance} "
    f"({highest_insurance_value:,.2f})"
)


# Longest stay condition
condition_stay = (
    df.groupby(
        "Medical Condition"
    )["Length of Stay"]
    .mean()
    .sort_values(
        ascending=False
    )
)

longest_stay_condition = (
    condition_stay.idxmax()
)

longest_stay_value = (
    condition_stay.max()
)

print(
    f"\n7. Medical condition with longest "
    f"average stay:"
)

print(
    f"   {longest_stay_condition} "
    f"({longest_stay_value:.2f} days)"
)


# Correlations
age_billing = (
    df["Age"]
    .corr(
        df["Billing Amount"]
    )
)

stay_billing = (
    df["Length of Stay"]
    .corr(
        df["Billing Amount"]
    )
)

print(
    f"\n8. Age vs Billing correlation:"
)

print(
    f"   {age_billing:.3f}"
)


print(
    f"\n9. Length of Stay vs Billing correlation:"
)

print(
    f"   {stay_billing:.3f}"
)


print("\n" + "=" * 70)


# ============================================================
# 46. FINAL KPI DATAFRAME
# ============================================================

kpi_table = pd.DataFrame({

    "Metric": [
        "Total Patient Records",
        "Average Age",
        "Median Age",
        "Average Billing",
        "Median Billing",
        "Total Recorded Billing",
        "Average Length of Stay",
        "Median Length of Stay",
        "Number of Hospitals",
        "Medical Conditions",
        "Insurance Providers",
        "Medications"
    ],

    "Value": [
        total_patients,
        round(average_age, 2),
        round(median_age, 2),
        round(average_billing, 2),
        round(median_billing, 2),
        round(total_billing, 2),
        round(average_stay, 2),
        round(median_stay, 2),
        unique_hospitals,
        unique_conditions,
        unique_insurance,
        unique_medications
    ]
})


print(
    "\n========== FINAL KPI TABLE =========="
)

display(kpi_table)


# ============================================================
# 47. SAVE CLEANED DATASET
# ============================================================

cleaned_file = (
    "hospital_healthcare_cleaned.csv"
)

df.to_csv(
    cleaned_file,
    index=False
)

print(
    "\n✅ Cleaned dataset saved:"
)

print(
    cleaned_file
)


# ============================================================
# 48. DOWNLOAD CLEANED DATASET
# ============================================================

files.download(
    cleaned_file
)


# ============================================================
# 49. FINAL PROJECT STATUS
# ============================================================

print("\n")
print("=" * 70)
print("🏥 HOSPITAL PATIENT & HEALTHCARE ANALYTICS")
print("PROJECT COMPLETED")
print("=" * 70)

print("""
✅ Dataset loaded
✅ Data quality assessment
✅ Missing-value analysis
✅ Duplicate analysis
✅ Data cleaning
✅ Date processing
✅ Length-of-stay calculation
✅ Patient demographic analysis
✅ Age-group analysis
✅ Gender analysis
✅ Medical condition analysis
✅ Admission analysis
✅ Admission trend analysis
✅ Length-of-stay analysis
✅ Billing analysis
✅ Insurance analysis
✅ Medication analysis
✅ Test-result analysis
✅ Hospital analysis
✅ Correlation analysis
✅ Statistical testing
✅ Automated findings
✅ KPI summary
✅ Clean dataset export


""")

print("=" * 70)
```
